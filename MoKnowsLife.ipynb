```python
def df_columns(df, phase):
    columns = [
        'Warm up', 'Repetition', 'Warm up', 'Repetition', 'Mobility',
        'Repetition', 'Cool Down', 'Repetition', 'Hips', 'Repetition',
        'Core', 'Repetition', 'Med Ball', 'Repetition', 'Sled (780m)',
        'Repetition', 'Plyo', 'Repetition', 'Posture', 'Repetition',
        'Sprint Reaction', 'Repetition', 'Wall Drills', 'Repetition'
    ]

    suffix_map = {
        "base": "1",
        "build": "2",
        "peak": "3"
    }

    df.columns = columns
    suffix = suffix_map.get(phase.lower())
    if suffix is None:
        raise ValueError("phase must be 'base', 'build', or 'peak'")

    new_df_columns = [col + suffix if col != 'Repetition' else col 
                      for col in columns]

    df.columns = new_df_columns
    return df


base_dff = df_columns(base_df, phase='base')
build_dff = df_columns(build_df, phase='build')
peak_dff = df_columns(peak_df, phase='peak')
```

```python



def split_repetitions(x):
    x = str(x).strip()

    # --- DISTANCE ONLY (m or yds) ---
    

    # --- SETS x REPS / DISTANCE / TIME ---
    if 'x' in x:
        left, right = x.split('x', 1)

        # left empty → "x2m"
        sets = int(left) if left else 1

        # right is distance
        if 'mtr' in right or 'yds' in right:
            distance = int(right.replace('yds', '').replace('mtr', ''))
            if 'yds' in right:
                distance = distance * 1.0936133
            return sets, 0, distance, 0

        # right is time
        if 'sec' in right or 'min' in right:
            if 'min' in right:
                duration = int(right.replace('min', '')) * 60
            else:
                duration = int(right.replace('sec', ''))
            return sets, 0, 0, duration

        # right is repetitions
        return sets, int(right), 0, 0

    if 'sec' in x or 'min' in x:
        if 'min' in x:
            duration = int(x.replace('min', '')) * 60
        else:
            duration = int(x.replace('sec', ''))
        return 0, 0, 0, duration
        
    if 'mtr' in x or 'yds' in x:
            distance = int(x.replace('yds', '').replace('mtr', ''))
            if 'yds' in x:
                distance = distance * 1.0936133
            return 0, 0, distance, 0
    
        # --- TIME ONLY (s, sec, min) ---
    
        
    # --- PURE REPETITIONS ---
    return 0, x, 0, 0

#sets,reps,distance,duration
split_repetitions('2x3 sec holds')
```
