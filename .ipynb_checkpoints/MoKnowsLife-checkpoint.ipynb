```python

def df_columns(df):
    # IMPROVEMENT: detect the phase from df.name if available
    name = getattr(df, "name", "").lower()

    columns = [
        'Warm up', 'Repetition', 'Warm up', 'Repetition', 'Mobility',
        'Repetition', 'Cool Down', 'Repetition', 'Hips', 'Repetition',
        'Core', 'Repetition', 'Med Ball', 'Repetition', 'Sled (780m)',
        'Repetition', 'Plyo', 'Repetition', 'Posture', 'Repetition',
        'Sprint Reaction', 'Repetition', 'Wall Drills', 'Repetition'
    ]

    new_df_columns = []

    for column in columns:
        if 'base' in name:
            col = column + '1'
            new_df_columns.append(col)
        elif 'build' in name:
            col = column + '2'
            new_df_columns.append(col)
        elif 'peak' in name:
            col = column + '3'
            new_df_columns.append(col)

    df.columns = new_df_columns
    return df

                
#final_indexes = [0,17,34]
base_dff = file_df.loc[1:16].reset_index().drop('index', axis=1)
base_df = base_dff[1:] 
base_df.columns = base_dff.iloc[0]
build_dff = file_df.loc[18:33].reset_index().drop('index', axis=1)
build_df = build_dff[1:]
build_df.columns = build_dff.iloc[0]
peak_dff = file_df.loc[35:50].reset_index().drop('index', axis=1)
peak_df = peak_dff[1:]
peak_df.columns = peak_dff.iloc[0]
peak_df
build_df
```
